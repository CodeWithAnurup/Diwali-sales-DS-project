CUSTOMER METRICS


```python
#RFM analysis (RFM = Recency, Frequency, Monetary)
rfm = df.groupby('User_ID').agg({
    'Orders': 'sum',
    'Amount': 'sum'
}).reset_index()

rfm.rename(columns={
    'Orders': 'Frequency',
    'Amount': 'Monetary'
}, inplace=True)

rfm.head()
```

Identify Customer Segments

```python
rfm['Monetary_Segment'] = pd.qcut(rfm['Monetary'], q=4, labels=['Low', 'Medium', 'High', 'VIP'])
rfm.head()
```

```python
rfm.groupby('Monetary_Segment')[['Frequency', 'Monetary']].mean()
```

K-Means to create clusters:


```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans


X = rfm[['Frequency', 'Monetary']]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

kmeans = KMeans(n_clusters=4, random_state=42)
rfm['Cluster'] = kmeans.fit_predict(X_scaled)

rfm.head()
```

```python
rfm.groupby('Cluster')[['Frequency', 'Monetary']].mean()
```

```python
import pandas as pd

# Create a copy of your RFM-Cluster data
cluster_summary = rfm.groupby('Cluster')[['Frequency', 'Monetary']].mean().reset_index()

# Round frequency & monetary
cluster_summary['Frequency'] = cluster_summary['Frequency'].round(1)
cluster_summary['Monetary'] = cluster_summary['Monetary'].round(0)

# Add Meaning column (based on cluster number)
meaning_map = {
    0: 'Low-value / Infrequent Buyers',
    1: 'VIP Customers — Highest Value',
    2: 'High-frequency & High-spending Customers',
    3: 'Mid-value Customers'
}

cluster_summary['Meaning'] = cluster_summary['Cluster'].map(meaning_map)

# Format monetary column with commas + rupee symbol
cluster_summary['Monetary'] = cluster_summary['Monetary'].apply(lambda x: f"₹{x:,.0f}")

cluster_summary
```
